## Library Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn. cluster import KMeans
from sklearn. decomposition import PCA
from sklearn.metrics import adjusted_rand_score
```

```python
df = pd.read_csv(r"C:\Users\rohit\Downloads\Skill Nexis Internship\week - 3\Practice\Iris.csv")
```

```python
df.sample(10)
```

```python
df["Species"].unique()
```

```python
df.isnull().sum()
```

```python
X=df.drop("Species",axis=1)
y=df["Species"]
```

## KMEANS 

```python
kmeans=KMeans(n_clusters=3)
kmeans.fit(X)
```

```python
kmeans.transform(X)
```

```python
kmeans.labels_
```

```python
print("The Cluster Centers are: ", kmeans.cluster_centers_)
```

```python
df["Predicted"]=kmeans.labels_
```

## Comparison

```python
print("Comparison: ", f'{adjusted_rand_score(y,df["Predicted"])*100:.3f}')
```

## PCA

```python
pca=PCA(n_components=2)
pca.fit(X)
```

```python
pca_transformed=pca.transform(X)
```

```python
df_reduced=pd.DataFrame(pca_transformed, columns=pca.get_feature_names_out())
```

```python
df_reduced["Target"]=kmeans.labels_
```

```python
df_reduced
```

## Visualizing the clusters

```python
sns.relplot(x="pca0", y="pca1", data=df_reduced, kind="scatter", hue="Target")
```

## Explained Variance

```python
v=pca.explained_variance_ratio_
print("Variance Explained: ",v[0]*100, v[1]*100 )
```
