```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
```

```python
df.sample(10)
```

```python
df_split=df[["total_bill","tip","size"]]
```

## Performing One Hot Encoding on "sex", "smoker", "time" columns

```python
onehot_encoder=OneHotEncoder()
```

```python
transformed_onehot=onehot_encoder.fit_transform(df[["sex","smoker","time"]]).toarray()
```

```python
transformed_onehot=np.array(transformed_onehot,dtype="int")
```

```python
col_names=onehot_encoder.get_feature_names_out(["sex","smoker","time"])
```

```python
new_dataset=pd.DataFrame(transformed_onehot,columns=col_names)
```

```python
transformed_dataset=pd.concat([df_split,new_dataset],axis=1)
```

## Performing Label Encoding on "day" column

```python
label_encoder=LabelEncoder()
```

```python
target=label_encoder.fit_transform(df["day"])
```

```python
transformed_dataset["day"]=target
```

```python
transformed_dataset.sample(10)
```
