```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
dataset=pd.read_csv(r"C:\Users\rohit\Downloads\Skill Nexis Internship\week-1\Assignments\house_price_chandigarh_dataset.xls")
```

```python
dataset.sample(10)
```

```python
dataset.shape[0]
```

```python
dataset.isnull().sum()
```

## Mean Imputation

```python
mean_imputer=SimpleImputer(missing_values=np.nan,strategy="mean",fill_value="constant")
```

```python
transformed_mean_imputation=mean_imputer.fit_transform(dataset)
```

```python
transformed_dataset_mean_imputation=pd.DataFrame(transformed_mean_imputation,columns=dataset.columns)
```

```python
transformed_dataset_mean_imputation.sample(10)
```

```python
transformed_dataset_mean_imputation.isnull().sum()
```

## Median Imputation

```python
median_imputer=SimpleImputer(missing_values=np.nan,strategy="most_frequent",fill_value="constant")
```

```python
transformed_median_imputation=median_imputer.fit_transform(dataset)
```

```python
transformed_dataset_median_imputation=pd.DataFrame(transformed_median_imputation,columns=dataset.columns)
```

```python
transformed_dataset_median_imputation.sample(10)
```

```python
transformed_dataset_median_imputation.isnull().sum()
```
